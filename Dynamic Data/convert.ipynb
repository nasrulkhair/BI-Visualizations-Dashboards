### Create Connection to save csv into SQL Server

```python
from sqlalchemy import create_engine
import pyodbc


# check for odbc driver
print(pyodbc.drivers())

# Create a connection to the SQL Server database
params = (
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=NASRULKHAIR\SQLEXPRESS;"
    "DATABASE=AdventureWorksDW;"
    "TRUSTED_CONNECTION=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")

excel_path = r"C:\Users\User\Desktop\Data Analyst\BI Visualizations & Dashboards\Dynamic Data\Power BI AW Dataset\AdventureWorksDW.xlsx"
file_dict = {
    'fact_internet_sales': fact_internet_sales, 
    'Dim_Product': Dim_Product,
    'Dim_SalesTerritory': Dim_SalesTerritory,
    'Dim_Date': Dim_Date,
    'Dim_Geography': Dim_Geography
}


for filename, df in file_dict.items():
    df.to_sql(filename, con=engine, if_exists='replace', index=False)
```
